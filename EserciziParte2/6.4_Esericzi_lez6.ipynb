# 4. Rispondere alle domande
Trasforma le seguenti richiesti in formule per estrarre infromazioni dal dataframe
- Mostrami le vendite maggiori di 10

- Mostrami i dati del 2018

- Mostrami le vendite maggiori di 13 e l'anno è il 2018

- Mostrami tutto TRANNE i casi in cui le vendite sono maggiori di 13 e l'anno è il 2018

- Mostrami i dati dove le vendite divise per 3 sono maggiori di 3

- Mostrami i dipendenti i cui nomi sono alfabeticamente dopo la J



```python
import pandas as pd

sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
sales
```

```python
sales.query('sales > 10')
```

```python
sales.query('year == 2018')
```

```python
sales.query("year == 2018 & sales > 13")
```

```python
sales.query("year != 2018 or sales < 13")
```

```python
def diviso3(numero):
    if(numero/3 > 3):
        return True
    else:
        return False
    
sales[sales["sales"].apply(diviso3) == 1]
```

```python
sales.query('employee > "J"')
```

# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
nomi = pd.unique(df['Name'])
len(nomi)
```

```python
df['Name'].count()
```

```python
result = df.groupby("Genre")["Global_Sales"].sum().reset_index()
genres = result["Genre"]
sales = result["Global_Sales"]

plt.bar(genres,sales)
plt.xticks(rotation =45)
plt.show
```

```python
result = df.groupby("Year")["Name"].count().reset_index()

result.plot(x="Year",y="Name")
plt.show()
```

```python

import numpy as np
generi = ['Action', 'Misc', 'Role-Playing', 'Shooter', 'Sports']
bottom = np.zeros(len(generi))
sales = ["NA_Sales","EU_Sales","JP_Sales","Other_Sales"]
myDict = {
    'NA_Sales' : [], 
    'EU_Sales': [], 
    'JP_Sales':[], 
    'Other_Sales':[], 
}
for i in generi:
    myDict['NA_Sales'].append(df[(df["Genre"] == i)]["NA_Sales"].sum())
    myDict['EU_Sales'].append(df[(df["Genre"] == i)]["EU_Sales"].sum())
    myDict['JP_Sales'].append(df[(df["Genre"] == i)]["JP_Sales"].sum())
    myDict['Other_Sales'].append(df[(df["Genre"] == i)]["Other_Sales"].sum())


fig, ax = plt.subplots()

for key, value in myDict.items():
    ax.bar(generi, value, bottom=bottom,label = key)
    #Questa cosa la puoi fare solo che Numpy, perchè se facevi la lista [0,0,0,0,0]
    #avresti dovuto modificare valore per valore
    bottom +=value

ax.legend(loc= "upper center")
plt.show()

```
